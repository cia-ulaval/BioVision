```python
import torch, cv2, ultralytics, fiftyone, pandas, numpy, os
from pathlib import Path



## Les images et vidéos seront stockées dans le dossier data 
## 1. Les vidéos sont dans un sous-dossier 'Videos'. 
## 2. Pour chacune des vidéos:
## - Créer un sous-dossier dans 'Images'
## - Lire et capturer la vidéo
## - Extraire les frames 
## - Enregistrer chaque frame extraite dans le sous-dossier créé
## 3. Répéter pour chacune des vidéos dans 'Videos'



def extraireimages(video_path: Path, sortie_dir: Path):      ## extraire les images de la vidéo

    sortie_dir.mkdir(parents=True, exist_ok=True)      ## créer un sous-dossier dans Images
    
    capture = cv2.VideoCapture(str(video_path))    ## comme pour "lire" une vidéo

   
    framecourante = 0               ## initialiser le compteur, début de la vidéo
    
    while capture.isOpened():   ## tant que la vidéo est ouverte

        ret,frame = capture.read()         ## capturer frame par frame

        if ret:                                 ## si la vidéo n'est pas finie
            name = sortie_dir / (str(framecourante)+'.jpg')   ## Donner un nom à la frame
                       
            if framecourante % 5== 0:
                cv2.imwrite(name,frame)
                
            framecourante = framecourante + 1

        else:
            break

    capture.release()   
    
    
    

def scriptextraction(dossier_video, dossier_sortie):

    ## utiliser une boucle for pour capturer et extraire des images de chacune des vidéos dans le sous-dossier 
    
    for video in Path(dossier_video).glob("*"):  ## Parcourir toutes les vidéos
        extraireimages(video, Path(dossier_sortie)/video.stem)   ## appeler la fonction d'extraction


    
```

```python
from pathlib import Path
scriptextraction("../data/Videos","../data/Images")
```
